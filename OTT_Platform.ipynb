```python
df.drop_duplicates(inplace=True)
```

```python
# Dirty Values - N/A , NULL, Blank, space

df.replace(
    ['N/A','NULL','',' '],np.nan,inplace=True
)
```

```python
df.isnull().sum()
```

```python
# Extra Spaces remove
df=df.apply(lambda x: x.str.strip() if x.dtype=="str" else x)
```

```python
df['Country'].value_counts()
```

```python
df['Title'].value_counts()
```

```python
# Standarize type

df["Content_Type"]=df["Content_Type"].str.title()
df["Customer_ID"]=df["Customer_ID"].str.title()
df["Customer_Name"]=df["Customer_Name"].str.title()
df["Title"]=df["Title"].str.title()
df["Genre"]=df["Genre"].str.title()
df["Country"]=df["Country"].str.title()
df["Subscription_Plan"]=df["Subscription_Plan"].str.title()
df["Device"]=df["Device"].str.title()
df["Payment_Method"]=df["Payment_Method"].str.title()
df["Account_Status"]=df["Account_Status"].str.title()
```

```python
# Numeric Columns

numeric_cols = ['Age', 'Watch_Hours','Monthly_Fee', 'Movies_Watched','Episodes_Watched', 'Avg_Rating','Monthly_Sessions', 'Download_Count']
```

```python
for col in numeric_cols:
    df[col]=pd.to_numeric(df[col],errors='coerce')
```

```python
df.describe()
```

```python
df=df[(df['Age']>=18)& (df['Age']<=100)]
```

```python
df['Age'].describe()
```

```python
# Negative Values

df=df[df['Watch_Hours']>=0]
df=df[df['Monthly_Fee']>=0]
```

```python
df.describe()
```

```python
df.shape
```

```python
# Date Cleaning

df['Join_Date']=pd.to_datetime(df['Join_Date'],errors='coerce')

df['Last_Watch_Date']=pd.to_datetime(df['Last_Watch_Date'],errors='coerce')
```

```python
df.info()
```

```python
df[["Join_Date","Last_Watch_Date"]].isnull().sum()
```

```python
# Missing Values

df['Age']=df['Age'].fillna(df["Age"].mean(), inplace=True)
df['Watch_Hours']=df['Watch_Hours'].fillna(df['Watch_Hours'].mean(), inplace=True)
df['Monthly_Fee']=df['Monthly_Fee'].fillna(df['Monthly_Fee'].mean(),inplace=True)
```

```python
df.describe()
```

```python
df['Payment_Method']=df['Payment_Method'].fillna("Unknown",inplace=True)
df['Device']=df['Device'].fillna("Unknown",inplace=True)
df['Genre']=df['Genre'].fillna("Unknown",inplace=True)
```

```python
# Feature Engineering

# Customer Revenue
df['Customer_Revenue']=df['Monthly_Fee'] * df['Monthly_Sessions']
```

```python
df.head()
```

```python
# Engagement Score

df['Engagement_Score']=(df['Watch_Hours']+df['Monthly_Sessions']+df['Download_Count'])
```

```python
df.head()
```

```python
# Watch Category 

bins=[0,20,50,100,np.inf]
label=['Low','Medium','High','Very High']

df['Watch_Category']=pd.cut(df['Watch_Hours'],bins=bins,labels=label)
```

```python
df.head()
```

```python
df['Watch_Category'].value_counts()
```

```python
# Customer Age Group

bins=[17,25,35,50,100]
label=['18-25','26-35','36-50','51+']

df["Age_Group"]=pd.cut(df['Age'],bins=bins,labels=label)
```

```python
df.head()
```

```python
df['Age_Group'].value_counts()
```

```python
# Account Activ Flag

df['Active_Flag']=np.where(df['Account_Status']=='Active',1,0)
```
